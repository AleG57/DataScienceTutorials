### Import des librairies

```python
from sklearn import datasets
from skimpy import skim
import pandas as pd
import plotly.express as px
from itertools import cycle
import numpy as np

from src.missingvalues import (
    missing_values_summary,
    missing_values_heatmap
)
```

### Import du *dataset*

```python
df = pd.read_csv("data/ETTh.csv")
```

```python
df.head()
```

```python
skim(df)
```

```python
features = df.select_dtypes(include=[np.number, 'timedelta64']).columns
cmaps_cycle = cycle([
    'Inferno',
    'Viridis',
    'Plasma',
    'Magma',
    'Turbo',
    'Hot', 
    'Electric'
])

for feature in features:
    current_cmap = next(cmaps_cycle)

    y_values = df[feature]
    if 'timedelta' in str(y_values.dtype):
        y_values = y_values.dt.total_seconds() / 3600
    

    fig = px.scatter(
        df,
        x='date',
        y=y_values,
        color=y_values,
        color_continuous_scale=current_cmap,
        title=f"Évolution de {feature}"
    )
    
    fig.update_traces(
        line=dict(width=2),
        connectgaps=False
    )
    
    fig.update_layout(
        #template="plotly_dark",
        xaxis_title="Chronologie",
        yaxis_title=feature,
        coloraxis_showscale=True,
        hovermode="x unified"
    )
    
    fig.show()
```
